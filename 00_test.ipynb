```python
# get marks of student and give a grade based on marks
# 95 and above - grade A
# 90 and above - grade B
# 80 and above - grade C
# 70 and above - grade D
# below 70 - grade F
marks = int(input("Enter your marks: "))
if marks >= 95:
    print("Grade A")
elif marks >= 90:
    print("Grade B")
elif marks >= 80:
    print("Grade C")
elif marks >= 70:
    print("Grade D")
else:
    print("Grade F")
```
